```python
import gradio as gr

import gradio as gr
from huggingface_hub import get_token

title = "Qwen2.5-7B-Instruct"
description = (
    "Gradio demo for Qwen2.5-7B-Instruct, served through Hugging Face Inference Providers. "
    "Type a message, or click one of the examples to load it."
)
examples = [
    "The tower is 324 metres (1,063 ft) tall,",
    "The Moon's orbit around Earth has",
    "The smooth Borealis basin in the Northern Hemisphere covers 40%",
]
gr.load(
    "models/Qwen/Qwen2.5-7B-Instruct",
    token=get_token(),
    title=title,
    description=description,
    examples=examples,
).launch()
```

```python
import gradio as gr
from gradio_client import Client, handle_file

client = Client("not-lain/background-removal")

def remove_background(image_path):
    original, processed = client.predict(image=handle_file(image_path), api_name="/image")
    return processed

gr.Interface(
    fn=remove_background,
    inputs=gr.Image(type="filepath", label="Input image"),
    outputs=gr.Image(label="Background removed"),
    title="Remove Background",
    description="Powered by the not-lain/background-removal Space.",
).launch()
```

```python
import gradio as gr
from gradio_client import Client, handle_file

client = Client("not-lain/background-removal")

def remove_background(image_path):
    original, processed = client.predict(image=handle_file(image_path), api_name="/image")
    return processed

gr.Interface(
    fn=remove_background,
    inputs=gr.Image(sources=["webcam"], type="filepath", label="Webcam"),
    outputs=gr.Image(label="Background removed"),
    title="Remove your webcam background!",
).launch()
```
